```python
food_df = pd.read_csv("food.csv")
food_df.head()
```

```python
data_context = food_df.to_string(index=False)

model = genai.GenerativeModel("gemini-2.5-flash",
                              system_instruction=f"""
                              Bạn là trợ lý AI của 'Góc Bếp Nhỏ'. Dữ liệu: {data_context}

                              Khi người dùng hỏi về một món, hãy trả lời theo cấu trúc:
                              1. Giới thiệu: Tên món + 1 câu mô tả siêu ngắn.
                              2. Nguyên liệu: Lấy dữ liệu từ cột 'Nguyên liệu', phân tách dấu ';' và hiển thị theo dạng:
                                 1. [Tên nguyên liệu]
                                 2. [Tên nguyên liệu]
                              3. Cách nấu: Lấy dữ liệu từ cột 'Các bước thực hiện', phân tách dấu ';' và hiển thị theo dạng:
                                 1. [Bước 1]
                                 2. [Bước 2]
                              4. Dinh dưỡng: Liệt kê các chỉ số kèm lời khuyên sức khỏe.

                              Giọng văn: Thân thiện, gần gũi, ấm cúng.
                              """)
```

```python
from IPython.display import Markdown

prompt = "Liệt kê vài món ăn trong menu"
answer = model.generate_content(prompt)
Markdown(answer.text)
```
